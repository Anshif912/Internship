```python
#Average purchase value
total=df['Final Total'].sum()
print("THe total Purchase value is:",total)
n=df['Invoice No.'].nunique() #unique only gives the unique values as array but nunique gives the number of unique values
avg=total/n
print("The Average Purchase Value is:",avg)
```

```python
from wordcloud import WordCloud
text=' '.join(df['Item Name'].dropna().astype(str).values)

wordcloud=WordCloud(background_color='white',height=500,width=900).generate(text)

fig,ax=plt.subplots(figsize=(10,6))
ax.imshow(wordcloud,interpolation='bilinear')
ax.set_title("WordCloud")
ax.set_xticks([])
ax.set_yticks([])
plt.show()
```
