```python
S=df.columns.tolist()
plt.figure(figsize=(30,10))
for i in range(1,len(S)):
  plt.subplot(3,3,i)
  plt.hist(df[S[i-1]],bins=30,edgecolor="black")
  plt.title(S[i-1])
plt.show()
```
